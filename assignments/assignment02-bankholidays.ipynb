# Northern Ireland Bank Holidays

```python
# Import the requests library
import requests 
```

## Retrieve bank holiday data

```python
# URL containing UK bank holiday data in JSON format
url = "https://www.gov.uk/bank-holidays.json"

# Send request and retrieve JSON response
response = requests.get(url)
data = response.json()
```

## All events in Northern Ireland

```python
# Extract the list of Northern Ireland events
# Events are stored as dictionaries inside a list
ni_events = data["northern-ireland"]["events"][:]

# Iterate through the list of events and display their titles and dates
event_count = 0
for event in ni_events:
    print(f' The date of {event["title"]} is {event["date"]}') 
    event_count += 1
```

## Events unique to Northern Ireland

### References

- **Iterating over dictionary keys**: The approach to looping through dictionary keys was inspired by a guide on [Towards Data Science](https://towardsdatascience.com/iterate-over-keys-and-values-python-dictionaries-34f0f01fa34d/) article.

- **Finding unique items between two sets**: TThe method for identifying unique events by subtracting one set from another follows the principles described in:
    - [Askpython](https://www.askpython.com/python/list/difference-between-two-lists-unique-entries)
    - [Geek for Geeks](https://www.geeksforgeeks.org/python/python-set-difference/)

```python
# Set to store all event titles from regions other than Northern Ireland
non_ni_only = set()

# Set to store Northern Ireland event titles
ni_only = set()

# Iterate through all regions in the dataset
for key in data.keys():
    
    # Collect events from all non–Northern Ireland regions
    if key != "northern-ireland":
        for event in data[key]['events']:
            # Add event title to the set of non-NI events
            non_ni_only.add(event["title"])
    
    # Collect Northern Ireland event titles
    else:
        for event in data[key]['events']:
            # Add event title to the set of NI only events
            ni_only.add(event["title"])
```

```python
# Subtract all non-NI event titles from NI event titles
# to determine events unique to Northern Ireland
unique_ni_holidays = ni_only - non_ni_only

print(f'The UK bank holidays unique to Northern Ireland are: {unique_ni_holidays}')
```
